## Dealing with bad records
1. PERMISSIVE
2. DROPMALFORMED
3. FAILFAST - if anyone record fail then it will not get pass to next step and fail itself


```python
spark.read.option("columnNameOfCorruptRecord", "bad_records").json("/Volumes/quickstart_catalog/quickstart_schema/sandbox/dataset/access_logs.json",  mode="PERMISSIVE").display()
```

## Dataframe Writer API
