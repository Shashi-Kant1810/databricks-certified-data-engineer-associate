```python
df = spark.read.csv(
    path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/dataset/user_dataset/users_001.csv",
    header=True,
    inferSchema=True,
)
```

## Create delta table

```python
%sql
USE CATALOG quickstart_catalog;
DROP TABLE IF EXISTS quickstart_schema.users;
CREATE TABLE IF NOT EXISTS quickstart_schema.users(
  id INT,
  name STRING,
  dob DATE,
  email STRING,
  gender STRING,
  country STRING,
  region STRING,
  city STRING,
  asset INT,
  marital_status STRING
);
DESCRIBE EXTENDED quickstart_schema.users;
```

## Data Loading 

```python
df.write.mode("overwrite").saveAsTable("quickstart_schema.users")
```

## Reading delta table

```python
spark.read.table("quickstart_schema.users").display()
```

## List Transaction history 

```python
from delta.tables import DeltaTable
deltatable=DeltaTable.forName(spark, "quickstart_schema.users")
deltatable.history().display()
```

## Describe using sql


```python
%sql
describe history quickstart_schema.users;
```

## Read specific version using sql

```python
%sql
select * from quickstart_schema.users version as of 0;
```

## update delta table

```python
%sql
UPDATE quickstart_schema.users
set country = 'Bharat'
where country = 'India'
```

```python
%sql
describe history quickstart_schema.users;
```

## customize userMetadata

```python
from pyspark.sql.functions import col
spark.read.csv(
    path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/dataset/user_dataset/users_001.csv",
    header=True,
    inferSchema=True,
).filter(col("city") == "Bangalore").write.option(
    "userMetadata", "Loading data from city='Bangalore'"
).mode(
    "overwrite"
).saveAsTable(
    "quickstart_schema.users"
)
```

```python
%sql
describe history quickstart_schema.users;
```
