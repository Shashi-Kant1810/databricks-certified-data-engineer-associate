```python
df = spark.read.csv(
    path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/dataset/user_dataset/users_001.csv",
    header=True,
    inferSchema=True,
)
```

## Transaction 01 - Write the data as delta format


```python
df.write.save(
    format="delta",
    mode="overwrite",
    path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/output_delta",
)
```

## Transaction 02 - Overwrite delta format

```python
from pyspark.sql.functions import col
df.filter(col("country")=="India").write.save(
    format="delta",
    mode="overwrite",
    path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/output_delta",
)
```

## Read data - always read latest data



```python
spark.read.format("delta").load(
    path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/output_delta"
).display()
```
